```python
# Forward Chaining - Observation Book Problem

# Initial Knowledge Base (Facts)
KB = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Missile", "T1"),
    ("Owns", "A", "T1")
}

# Forward Chaining Algorithm
def forward_chaining(KB):

    inferred = True

    while inferred:
        inferred = False
        new_facts = set()

        # Rule 1: Enemy(x, America) -> Hostile(x)
        for fact in KB:
            if fact[0] == "Enemy" and fact[2] == "America":
                new_fact = ("Hostile", fact[1])
                if new_fact not in KB:
                    new_facts.add(new_fact)

        # Rule 2: Missile(x) -> Weapon(x)
        for fact in KB:
            if fact[0] == "Missile":
                new_fact = ("Weapon", fact[1])
                if new_fact not in KB:
                    new_facts.add(new_fact)

        # Rule 3: Missile(x) AND Owns(y,x)
        #         -> Sells(y,x,A)
        for m in KB:
            if m[0] == "Missile":
                x = m[1]

                for o in KB:
                    if o[0] == "Owns" and o[2] == x:
                        new_fact = ("Sells", "Robert", x, o[1])
                        if new_fact not in KB:
                            new_facts.add(new_fact)

        # Rule 4:
        # American(p) AND Weapon(q) AND
        # Sells(p,q,r) AND Hostile(r) -> Criminal(p)

        for a in KB:
            if a[0] == "American":
                p = a[1]

                for w in KB:
                    if w[0] == "Weapon":
                        q = w[1]

                        for s in KB:
                            if (s[0] == "Sells" and
                                s[1] == p and s[2] == q):

                                r = s[3]

                                for h in KB:
                                    if h == ("Hostile", r):
                                        new_fact = ("Criminal", p)

                                        if new_fact not in KB:
                                            new_facts.add(new_fact)

        # Add newly inferred facts to KB
        if new_facts:
            print("\nNew facts inferred:")
            for fact in new_facts:
                print(fact)
            KB.update(new_facts)
            inferred = True

    return KB


# Execute Forward Chaining
result = forward_chaining(KB)

# Display final Knowledge Base
print("\nFinal Knowledge Base:")
for fact in sorted(result):
    print(fact)

# Check the goal
goal = ("Criminal", "Robert")

print("\nGoal:")
if goal in result:
    print("Robert is Criminal.")
    print("hence proved")
else:
    print("Cannot prove that Robert is Criminal.")




print("Name : AMULYA")
print("USN: 1WN24CS036")

```
