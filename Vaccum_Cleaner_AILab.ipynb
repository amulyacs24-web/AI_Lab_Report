Vaccum Cleaner Agent

```python
def vacuum_cleaner_agent(location, status):
    """Simple reflex agent for the vacuum cleaner problem."""
    print(f"Agent is located in Room: {location}")
    print(f"Room Status: {status}")

    # Action rules based on percepts
    if status[location] == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    elif location == "B":
        return "Move Left"
    else:
        return "NoOp"  # No operation if already clean and positioned


# Initialize environment state
environment = {"A": "Dirty", "B": "Dirty"}

# Initial agent location
agent_location = "A"

# Simulate 4 steps of the vacuum cleaner agent
for step in range(1, 5):
    print(f"\n--- Step {step} ---")
    action = vacuum_cleaner_agent(agent_location, environment)
    print(f"Action chosen: {action}")

    if action == "Suck":
        environment[agent_location] = "Clean"
        print(f"Room {agent_location} has been cleaned.")
    elif action == "Move Right":
        agent_location = "B"
        print("Moved to Room B.")
    elif action == "Move Left":
        agent_location = "A"
        print("Moved to Room A.")

    print(f"Updated Environment: {environment}")
```
