```python
import heapq
import time

GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

def find_position(state, value):
    for r in range(3):
        for c in range(3):
            if state[r][c] == value:
                return r, c

def manhattan_distance(state):
    distance = 0
    for r in range(3):
        for c in range(3):
            value = state[r][c]
            if value != 0:
                target_r, target_c = divmod(value - 1, 3)
                distance += abs(r - target_r) + abs(c - target_c)
    return distance

def get_neighbors(state):
    neighbors = []
    r, c = find_position(state, 0)
    moves = [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]
    for dr, dc, direction in moves:
        new_r, new_c = r + dr, c + dc
        if 0 <= new_r < 3 and 0 <= new_c < 3:
            new_board = [list(row) for row in state]
            new_board[r][c], new_board[new_r][new_c] = new_board[new_r][new_c], new_board[r][c]
            neighbor_state = tuple(tuple(row) for row in new_board)
            neighbors.append((neighbor_state, direction))
    return neighbors

def solve_8_puzzle(start_state):
    start_time = time.time()
    start_h = manhattan_distance(start_state)
    open_set = [(start_h, 0, start_state, [])]
    visited = set()

    nodes_expanded = 0
    max_memory_states = 1
    while open_set:

        current_space = len(open_set) + len(visited)
        if current_space > max_memory_states:
            max_memory_states = current_space

        f, g, current_state, path = heapq.heappop(open_set)

        if current_state == GOAL_STATE:
            execution_time = time.time() - start_time
            return path, nodes_expanded, max_memory_states, execution_time

        if current_state in visited:
            continue

        visited.add(current_state)
        nodes_expanded += 1
        for neighbor, direction in get_neighbors(current_state):
            if neighbor not in visited:
                new_g = g + 1
                new_h = manhattan_distance(neighbor)
                new_f = new_g + new_h
                heapq.heappush(open_set, (new_f, new_g, neighbor, path + [direction]))

    execution_time = time.time() - start_time
    return None, nodes_expanded, max_memory_states, execution_time

def print_state(state):
    for row in state:
        print(" ".join(str(tile) for tile in row))
    print()

if __name__ == "__main__":
    initial_puzzle = (
        (1, 2, 3),
        (0, 4, 6),
        (7, 5, 8)
    )

    print("Initial State:")
    print_state(initial_puzzle)

    solution_path, time_complexity, space_complexity, duration = solve_8_puzzle(initial_puzzle)

    if solution_path is not None:
        print(f" Solution found in {len(solution_path)} moves!")
        print("Steps to solve:", " -> ".join(solution_path))
        print("-" * 40)
        print("COMPLEXITY ANALYSIS:")
        print(f"• Time Complexity (Nodes Expanded): {time_complexity} nodes")
        print(f"• Space Complexity (Max Active States): {space_complexity} states")
        print(f"• Actual Execution Time: {duration:.6f} seconds")
        print("-" * 40)
    else:
        print("❌ This puzzle configuration is unsolvable.")

    print("Amulya 1WN24CS036")
```
