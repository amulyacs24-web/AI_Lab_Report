```python
import random

def heuristic(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts

def print_board(board):
    n = len(board)

    for i in range(n):
        for j in range(n):
            if board[j] == i:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

def hill_climbing(n):
    board = [random.randint(0, n - 1) for _ in range(n)]

    print("Initial Board:")
    print_board(board)
    print("Heuristic:", heuristic(board))

    step = 1

    while True:
        current_h = heuristic(board)

        if current_h == 0:
            return board

        best_board = board[:]
        best_h = current_h

        for col in range(n):
            for row in range(n):
                if row != board[col]:
                    new_board = board[:]
                    new_board[col] = row
                    h = heuristic(new_board)

                    if h < best_h:
                        best_h = h
                        best_board = new_board

        if best_h >= current_h:
            return None

        print("\nStep", step)
        print("Queen moved:")
        for i in range(n):
            if board[i] != best_board[i]:
                print("Column", i + 1, ": Row", board[i] + 1, "-> Row", best_board[i] + 1)

        board = best_board

        print("New Board:")
        print_board(board)
        print("Heuristic:", heuristic(board))

        step += 1

n = int(input("Enter the value of N: "))

solution = hill_climbing(n)

if solution:
    print("\nFinal Solution:")
    print_board(solution)
    print("Heuristic:", heuristic(solution))
else:
    print("\nLocal minimum reached. No solution found.")
```

```python
import random
import math

def heuristic(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts

def simulated_annealing(n):
    while True:
        board = [random.randint(0, n - 1) for _ in range(n)]
        temperature = 100
        cooling_rate = 0.95

        while temperature > 0.01:
            current_h = heuristic(board)

            if current_h == 0:
                return board

            new_board = board[:]
            col = random.randint(0, n - 1)
            row = random.randint(0, n - 1)

            while row == board[col]:
                row = random.randint(0, n - 1)

            new_board[col] = row

            new_h = heuristic(new_board)
            delta = new_h - current_h

            if delta < 0 or random.random() < math.exp(-delta / temperature):
                board = new_board

            temperature *= cooling_rate

def print_board(board):
    n = len(board)

    for i in range(n):
        for j in range(n):
            if board[j] == i:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

n = 8
solution = simulated_annealing(n)

print("Solution found:")
print_board(solution)
print("Heuristic:", heuristic(solution))
```
